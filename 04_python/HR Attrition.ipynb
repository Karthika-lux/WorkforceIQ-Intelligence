```python
def summarize(col):
    t = df.groupby(col, observed=True).agg(
        headcount=("EmployeeNumber", "count"),
        employees_left=("Attrition_Flag", "sum")
    ).reset_index()

    t["active_employees"] = t["headcount"] - t["employees_left"]

    t["attrition_rate_pct"] = (
        100 * t["employees_left"] / t["headcount"]
    ).round(2)

    return t
```

## Attrition Rate by Department.

```python
def plot_rate(t, col, title, filename, labels=None,
              horizontal=False, figsize=(8, 4.5)):

    names = t[col].astype(str).tolist() if labels is None else [
        labels.get(x, str(x)) for x in t[col]
    ]

    rates = t["attrition_rate_pct"].values
    colors = [ORANGE if r > overall_rate else BLUE for r in rates]

    fig, ax = plt.subplots(figsize=figsize)

    if horizontal:
        bars = ax.barh(names, rates, color=colors)
        ax.axvline(overall_rate, color="grey", linestyle="--")
        ax.invert_yaxis()
        ax.set_xlabel("Attrition rate (%)")

        for b, r, n in zip(bars, rates, t["headcount"]):
            ax.text(r + 0.5, b.get_y() + b.get_height()/2,
                    f"{r:.1f}% (n={n})", va="center", fontsize=9)

    else:
        bars = ax.bar(names, rates, color=colors)
        ax.axhline(overall_rate, color="grey", linestyle="--")
        ax.set_ylabel("Attrition rate (%)")

        for b, r, n in zip(bars, rates, t["headcount"]):
            ax.text(b.get_x() + b.get_width()/2, r + 0.6,
                    f"{r:.1f}%\n(n={n})", ha="center", fontsize=9)

    ax.set_title(title, fontsize=12, fontweight="bold")
    sns.despine()
    plt.tight_layout()

    plt.savefig(f"C:/Users/dell/charts/{filename}", dpi=150)
    plt.show()
    plt.close()
```

```python
def plot_rate(t, col, title, filename, labels=None,
              horizontal=False, figsize=(8, 4.5)):

    names = t[col].astype(str).tolist() if labels is None else [
        labels.get(x, str(x)) for x in t[col]
    ]

    rates = t["attrition_rate_pct"].values
    colors = [ORANGE if r > overall_rate else BLUE for r in rates]

    fig, ax = plt.subplots(figsize=figsize)

    if horizontal:
        bars = ax.barh(names, rates, color=colors)
        ax.axvline(overall_rate, color="grey", linestyle="--")
        ax.invert_yaxis()
        ax.set_xlabel("Attrition rate (%)")

        for b, r, n in zip(bars, rates, t["headcount"]):
            ax.text(
                r + 0.5,
                b.get_y() + b.get_height() / 2,
                f"{r:.1f}% (n={n})",
                va="center",
                fontsize=9
            )

        ax.set_xlim(0, max(rates) * 1.25)

    else:
        bars = ax.bar(names, rates, color=colors)
        ax.axhline(overall_rate, color="grey", linestyle="--")
        ax.set_ylabel("Attrition rate (%)")

        for b, r, n in zip(bars, rates, t["headcount"]):
            ax.text(
                b.get_x() + b.get_width() / 2,
                r + 0.6,
                f"{r:.1f}%\n(n={n})",
                ha="center",
                fontsize=9
            )

        ax.set_ylim(0, max(rates) * 1.25)

    ax.set_title(title, fontsize=12, fontweight="bold")
    sns.despine()
    plt.tight_layout()

    plt.savefig(f"../06_docs/charts/{filename}", dpi=150)
    plt.show()
    plt.close()
```

## segments function

```python
def segments(cols, min_size=30, top=8):

    t = (
        df.groupby(cols, observed=True)
        .agg(
            headcount=("EmployeeNumber", "count"),
            employees_left=("Attrition_Flag", "sum")
        )
        .reset_index()
    )

    t["attrition_rate_pct"] = (
        100 * t["employees_left"] / t["headcount"]
    ).round(2)

    t = t[t["headcount"] >= min_size]

    t = t.sort_values(
        "attrition_rate_pct",
        ascending=False
    )

    return t.head(top)
```

## Attrition Summary Data

```python
summary_columns = [
    "Department",
    "JobRole",
    "Age_Group",
    "Tenure_Group",
    "OverTime",
    "Salary_Band",
    "JobSatisfaction",
    "WorkLifeBalance"
]

summary_list = []

for col in summary_columns:
    temp = (
        df.groupby(col, observed=True)
        .agg(
            headcount=("EmployeeNumber", "count"),
            employees_left=("Attrition_Flag", "sum")
        )
        .reset_index()
    )

    temp["attrition_rate_pct"] = (
        temp["employees_left"] / temp["headcount"] * 100
    ).round(2)

    temp["group_name"] = col
    temp = temp.rename(columns={col: "group_value"})

    summary_list.append(temp)

attrition_summary = pd.concat(summary_list, ignore_index=True)

attrition_summary.head()
```
